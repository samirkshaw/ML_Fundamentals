## # Matplotlib challenge - linear regression visualization with noise

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,50)
y = 3*x + np.random.rand(50)*2
y_pred = 3*x
```

```python
fig , (plot1,plot2) = plt.subplots(1,2, figsize=(12,5))

plot1.scatter(x,y,color="blue")
plot1.set_title("Raw Data")
plot2.scatter(x,y,color="blue",label ="Actual")
plot2.plot(x,y_pred,color="red", label ="Model")
plot2.set_title("Model Prediction")
plot2.legend()
plt.show()
```
