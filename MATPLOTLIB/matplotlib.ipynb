## # Matplotlib basics - line plot, scatter, subplots, legend, labels

```python
import matplotlib.pyplot as plt
print("matplotlib ready")
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x,y)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.title("My First Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, color="red", linewidth=3)
plt.title("My First Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2.1, 3.8, 6.2, 7.9, 10.3]

plt.scatter(x,y, color="blue")
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2.1, 3.8, 6.2, 7.9, 10.3]

y_pred = [2, 4, 6, 8, 10]  # perfect 2x line

plt.scatter(x, y, color="blue", label="Actual data")
plt.plot(x, y_pred, color="red", label="Model prediction")
plt.title("Data vs Model")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.plot(x, y1, color="red")
ax1.set_title("Linear y=2x")

ax2.plot(x, y2, color="green")
ax2.set_title("Quadratic y=x²")

plt.show()
```
