```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

x_train = np.array([600, 800, 1000, 1200, 1400, 1600, 1800, 2000, 2400, 3000]) #in sq feet
y_train = np.array([6, 8, 10, 12, 18, 22, 26, 30, 40, 55]) # in lacks
```

```python
plt.plot(x_train, y_train, )
plt.xlabel('Size of the house (sq ft)')
plt.ylabel('Price of the house ( in lakhs)')
plt.title('Dataset: Size vs Price')
plt.show()
```

### Linear regression


```python
w=0
b=0
y_cap = w * x_train + b
```

```python
y_diff = 0
for i in range(len(y_train)):
    y_diff += (y_train[i] - y_cap[i])**2
J = y_diff/(2*len(y_train))
print(J)
```

```python
dJ_dw = 0
for i in range(len(y_train)):
    dJ_dw += (y_cap[i] - y_train[i]) * x_train[i]
dJ_dw = -dJ_dw/len(y_train)
print(dJ_dw)

dJ_db = 0
for i in range(len(y_train)):
    dJ_db += (y_train[i] - y_cap[i])
dJ_db = -dJ_db/len(y_train)
print(dJ_db)
```

### Gradient Descent Algorithm

```python
x_mean, x_std = x_train.mean(), x_train.std()
x_norm = (x_train - x_mean) / x_std

w, b = 0.0, 0.0
alpha = 0.03
iterations = 500
m = len(x_train)

cost_history = []

for i in range(iterations):
    y_cap = w * x_norm + b
    dJ_dw = (1/m) * np.sum((y_cap - y_train) * x_norm)
    dJ_db = (1/m) * np.sum(y_cap - y_train)
    w -= alpha * dJ_dw
    b -= alpha * dJ_db
    cost_history.append((1/(2*m)) * np.sum((y_cap - y_train)**2))

print(f"w = {w:.4f}, b = {b:.4f}")
print(f"final cost = {cost_history[-1]:.4f}")

plt.plot(cost_history)
plt.title("Cost vs Iterations")
plt.xlabel("Iteration")
plt.ylabel("Cost J")
plt.show()
```

```python
x_new = int(input("Enter the size of the house in sq ft: "))
x_new_norm = (x_new - x_mean) / x_std
y = w * x_new_norm + b
print(f"The predicted price of the house is: {y:.2f} lakhs")
```
